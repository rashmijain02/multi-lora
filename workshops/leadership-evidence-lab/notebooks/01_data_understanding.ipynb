# 01 - Understand fictional support data

Original fictional teaching exchanges only. No EPITOME, Reddit, or TalkLife text is loaded. Labels illustrate emotional reactions, interpretations, and explorations; they do not measure anyone's personality or reproduce the published study.

Select the workshop environment as your kernel. Start from the repository root, workshop directory, or this notebooks directory.

```python
import json
from collections import Counter
from pathlib import Path

candidates = [Path.cwd(), *Path.cwd().parents]
ROOT = next((p for p in candidates if (p / 'data' / 'fictional_support.jsonl').is_file()), None)
if ROOT is None:
    ROOT = next((p / 'workshops' / 'leadership-evidence-lab' for p in candidates
                 if (p / 'workshops' / 'leadership-evidence-lab' / 'data' / 'fictional_support.jsonl').is_file()), None)
if ROOT is None:
    raise FileNotFoundError('Open this notebook inside the multi-lora repository.')
from empathy_lab.data import load_dataset, split_records

records = load_dataset(ROOT / 'data' / 'fictional_support.jsonl')
print('Fictional records:', len(records))
print('Record fields:', sorted(records[0]))
```

## Read before modeling

Inspect the record fields and a few fictional exchanges below. For each mechanism, `0` means absent, `1` limited, and `2` explicit/strong expression under this simplified teaching rubric. The original paper's annotation procedure is not being reproduced.

Rationales are annotation evidence in the response. The initial generative task predicts labels only, not rationale extraction. Ask whether a different reasonable annotator could disagree.

```python
for record in records[:3]:
    print(json.dumps(record, indent=2, ensure_ascii=False))
```

```python
print('Split counts:', dict(Counter(record['split'] for record in records)))
splits = split_records(records)
for split, rows in splits.items():
    print(split, {mechanism: dict(Counter(row['labels'][mechanism]['score'] for row in rows))
                  for mechanism in records[0]['labels']})
    print('Scenario families:', sorted({row['scenario_family'] for row in rows}))
print('Scenario and label summaries should be interpreted as fixture coverage, not population statistics.')
print('Total source bytes:', (ROOT / 'data' / 'fictional_support.jsonl').stat().st_size)
```
